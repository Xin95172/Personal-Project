```python
import subprocess
import sys

# 在後台執行 get_odds.py 腳本（因為它是無限循環的）
process = subprocess.Popen([sys.executable, 'get_odds.py'])

# 顯示進程ID
print(f"get_odds.py 已啟動，進程ID: {process.pid}")
print("腳本正在後台運行，按 Ctrl+C 可停止（如果在終端中運行）")
```

```python
# 停止 get_odds.py 進程（如果需要）
if 'process' in locals() and process.poll() is None:
    process.terminate()
    print("get_odds.py 已停止")
else:
    print("沒有運行中的 get_odds.py 進程")
```
