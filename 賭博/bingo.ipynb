```python
import pandas as pd
import numpy as np

from scipy.stats import hypergeom
```

```python
# 平常
# multiplier = {
#     1: {1: 2},
#     2: {2: 3, 1: 1},
#     3: {3: 20, 2: 2},
#     4: {4: 40, 3: 4, 2: 1},
#     5: {5: 300, 4: 20, 3: 2},
#     6: {6: 1000, 5: 40, 4: 8, 3: 1},
#     7: {7: 3200, 6: 120, 5: 12, 4: 2, 3: 1},
#     8: {8: 20000, 7: 800, 6: 40, 5: 8, 4: 1, 0: 1},
#     9: {9: 40000, 8: 4000, 7: 120, 6: 20, 5: 4, 4: 1, 0: 1},
#     10: {10: 200000, 9: 10000, 8: 1000, 7: 100, 6: 10, 5: 1, 0: 1}
# }

# 加碼
multiplier = {
    1: {1: 3},
    2: {2: 6, 1: 1},
    3: {3: 40, 2: 2},
    4: {4: 80, 3: 6, 2: 1},
    5: {5: 400, 4: 24, 3: 2},
    6: {6: 2000, 5: 48, 4: 8, 3: 1},
    7: {7: 3200, 6: 120, 5: 12, 4: 2, 3: 1},
    8: {8: 20000, 7: 800, 6: 40, 5: 8, 4: 1, 0: 1},
    9: {9: 40000, 8: 4000, 7: 120, 6: 20, 5: 4, 4: 1, 0: 1},
    10: {10: 200000, 9: 10000, 8: 1000, 7: 100, 6: 10, 5: 1, 0: 1}
}
```

```python
N, n = 80, 20
x_vals = np.arange(0, 11)
stars_list = sorted(multiplier.keys())

# 1) 機率矩陣 P
P = pd.DataFrame(0.0, index=x_vals, columns=stars_list)
for K in stars_list:
    P.loc[x_vals, K] = hypergeom.pmf(x_vals, N, K, n)

# 2) 倍數矩陣 M
M = pd.DataFrame(0.0, index=x_vals, columns=stars_list)
for s, pay in multiplier.items():
    for x, m in pay.items():
        M.loc[x, s] = m
M = M - 1 # 扣掉本金

# 3) 期望倍數（每列）
EV = (P * M).sum(axis=0)

pd.options.display.float_format = "{:.6f}".format
display(P)       # 機率表
display(M)       # 倍數表
display(EV)      # 期望倍數
```
