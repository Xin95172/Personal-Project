```python
# get training set
dtm_csr_BoW_path = "../lexicon_resources/dtm_csr_BoW.npz"
dtm_csr_BoW = sp.load_npz(dtm_csr_BoW_path)

dtm_csr_TF_IDF_path = "../lexicon_resources/dtm_csr_TF_IDF.npz"
dtm_csr_TF_IDF = sp.load_npz(dtm_csr_TF_IDF_path)

verdict_results_path = "../artifacts/reports/verdict_results.xlsx"
verdict_results = pd.read_excel(verdict_results_path, index_col=0).to_numpy()

# BoW
x_train_bow, x_temp_bow, y_train_bow, y_temp_bow = train_test_split(
    dtm_csr_BoW, verdict_results, test_size=0.3, random_state=42
)
x_val_bow, x_test_bow, y_val_bow, y_test_bow = train_test_split(x_temp_bow, y_temp_bow, test_size=0.66666, random_state=42)

y_train_bow, y_val_bow, y_test_bow = np.asarray(y_train_bow).ravel(), np.asarray(y_val_bow).ravel(), np.asarray(y_test_bow).ravel()

print("x_train_bow:", x_train_bow.shape)
print("x_val_bow:", x_val_bow.shape)
print("x_test_bow:", x_test_bow.shape)
print("y_train_bow:", y_train_bow.shape)
print("y_val_bow:", y_val_bow.shape)
print("y_test_bow:", y_test_bow.shape)

# Tf-Idf
x_train_tfidf, x_temp_tfidf, y_train_tfidf, y_temp_tfidf = train_test_split(
    dtm_csr_BoW, verdict_results, test_size=0.3, random_state=42
)
x_val_tfidf, x_test_tfidf, y_val_tfidf, y_test_tfidf = train_test_split(x_temp_tfidf, y_temp_tfidf, test_size=0.66666, random_state=42)
```

### MNIR

```python
feature_res = load_or_fit_feature_splits(
    X_train=x_train_bow,
    y_train=y_train_bow,
    X_val=x_val_bow,
    X_test=x_test_bow,
    train_name="mnir_z_train_bow.npy",
    val_name="mnir_z_val_bow.npy",
    test_name="mnir_z_test_bow.npy",
    model_name="mnir_mnlm_bow.rds",
)

z_train_bow = feature_res["z_train"]
z_val_bow = feature_res["z_val"]
z_test_bow = feature_res["z_test"]

print("[MNIR] loaded_from_cache:", feature_res["loaded_from_cache"])
print("x_train_bow:", x_train_bow.shape)
print("x_val_bow:", x_val_bow.shape)
print("x_test_bow:", x_test_bow.shape)
print("z_train_bow:", z_train_bow.shape)
print("z_val_bow:", z_val_bow.shape)
print("z_test_bow:", z_test_bow.shape)

svm_grid = [
    {"C": 0.0625, "max_iter": 10000},
    {"C": 0.125, "max_iter": 10000},
    {"C": 0.25, "max_iter": 10000},
    {"C": 0.5, "max_iter": 10000},
    {"C": 1.0, "max_iter": 10000},
    {"C": 2.0, "max_iter": 10000},
    {"C": 4.0, "max_iter": 10000},
]

results = []
best_model = None
best_pred = None
best_score = -np.inf
best_params = None

for params in svm_grid:
    clf = SVMClassifier(**params)
    clf.fit(z_train_bow, y_train_bow)
    y_pred = clf.predict(z_val_bow)

    acc = accuracy_score(y_val_bow, y_pred)
    macro_f1 = f1_score(y_val_bow, y_pred, average="macro")
    weighted_f1 = f1_score(y_val_bow, y_pred, average="weighted")

    results.append({
        "C": params["C"],
        "max_iter": params["max_iter"],
        "accuracy": acc,
        "macro_f1_mean": macro_f1,
        "weighted_f1": weighted_f1,
    })

    if macro_f1 > best_score:
        best_score = macro_f1
        best_model = clf
        best_pred = y_pred
        best_params = params

tuning_report = pd.DataFrame(results).sort_values("macro_f1_mean", ascending=False)
display(tuning_report)

print("best params:", best_params)
print("best macro_f1_mean:", round(best_score, 4))

report_df = pd.DataFrame(
    classification_report(y_val_bow, best_pred, output_dict=True, zero_division=0)
).T
display(report_df)
```
