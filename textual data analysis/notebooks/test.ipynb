```python
import matplotlib.pyplot as plt
import numpy as np

# 設置中文字體 (若環境不支持可改英文)
plt.rcParams['font.sans-serif'] = ['Arial Unicode MS'] # 針對 Mac
plt.rcParams['axes.unicode_minus'] = False

labels = np.array(['Win F1', 'Loss F1', 'Partial F1', 'Weighted Avg', 'Overall Acc'])
stats_bow = np.array([0.73, 0.66, 0.10, 0.64, 0.68])
stats_tfidf = np.array([0.80, 0.76, 0.21, 0.73, 0.77])

angles = np.linspace(0, 2*np.pi, len(labels), endpoint=False).tolist()

# 閉合多邊形
stats_bow = np.concatenate((stats_bow, [stats_bow[0]]))
stats_tfidf = np.concatenate((stats_tfidf, [stats_tfidf[0]]))
angles = np.concatenate((angles, [angles[0]]))

fig, ax = plt.subplots(figsize=(8, 8), subplot_kw=dict(polar=True))

# 畫 BoW
ax.plot(angles, stats_bow, 'o-', linewidth=2, label='BoW + MNLR', color='#8ebad9')
ax.fill(angles, stats_bow, alpha=0.25, color='#8ebad9')

# 畫 TF-IDF
ax.plot(angles, stats_tfidf, 'o-', linewidth=2, label='TF-IDF + MNLR', color='#ffb347')
ax.fill(angles, stats_tfidf, alpha=0.25, color='#ffb347')

ax.set_thetagrids(np.degrees(angles[:-1]), labels)
ax.set_ylim(0, 1)
ax.set_title("MNLR Model Performance Comparison: BoW vs TF-IDF", va='bottom', fontsize=15)
ax.legend(loc='upper right', bbox_to_anchor=(1.3, 1.1))

plt.show()
```
