```python
df = pd.read_excel("xin.xlsx")
NA_CDS_df, EUR_CDS_df, emerging_CDS_df, e_u_df, cl1_comdty_df, vix_df, spx_df = df.iloc[1: , 0: 5].rename(columns = {"Date": "date"}).dropna(), df.loc[1: , "Date.1": "賣價"].rename(columns = {"Date.1": "date"}).dropna(), df.loc[1: , "Date.2": "賣價.1"].rename(columns = {"Date.2": "date"}).dropna(), df.loc[1: , "Date.3": "最新價"].rename(columns = {"Date.3": "date"}).dropna(), df.loc[1: , "Date.4": "SMAVG (15)"].rename(columns = {"Date.4": "date"}).dropna(), df.loc[1: , "Date.5": "最新價.2"].rename(columns = {"Date.5": "date"}).dropna(), df.loc[1: , "Date.6": "最新價.3"].rename(columns = {"Date.6": "date"}).dropna()
```

# SPX - CDS, VIX

```python
cl1_comdty_df
```

```python
df = spx_df.merge(
    NA_CDS_df,
    how = "left",
    on = "date"
)
df = df.merge(
    vix_df,
    how = "left",
    on = "date"
)
df = df.merge(
    EUR_CDS_df,
    how = "left",
    on = "date"
)
df = df.merge(
    emerging_CDS_df,
    how = "left",
    on = "date"
)
df = df.merge(
    e_u_df,
    how = "left",
    on = "date"
)
df = df.merge(
    cl1_comdty_df,
    how = "left",
    on = "date"
)
df = df.rename(columns = {
    "最新價.3": "SPX",
    "收盤": "NA_CDS",
    "最新價.2": "VIX",
    "賣價": "EUR_CDS",
    "賣價.1": "emerging_risk_prc",
    "最新價": "E_U_exchange",
    "最新價.1": "oil"
})
```

```python
plot(df, "SPX", x = "date", ry = "NA_CDS", ry_dashed = False)
plot(df, "SPX", x = "date", ry = "VIX", ry_dashed = False)
```
