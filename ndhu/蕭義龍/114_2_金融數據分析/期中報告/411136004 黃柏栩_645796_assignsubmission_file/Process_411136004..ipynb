```
import numpy as np
import IPython.display as display
from matplotlib import pyplot as plt
import io
import base64

ys = 200 + np.random.randn(100)
x = [x for x in range(len(ys))]

fig = plt.figure(figsize=(4, 3), facecolor='w')
plt.plot(x, ys, '-')
plt.fill_between(x, ys, 195, where=(ys > 195), facecolor='g', alpha=0.6)
plt.title("Sample Visualization", fontsize=10)

data = io.BytesIO()
plt.savefig(data)
image = F"data:image/png;base64,{base64.b64encode(data.getvalue()).decode()}"
alt = "Sample Visualization"
display.display(display.Markdown(F"""![{alt}]({image})"""))
plt.close(fig)
```

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import scipy.optimize as sco

# --- 1. 讀取與準備資料 ---
# 確保檔名與您上傳至 Colab 左側目錄的名稱完全一致
filename = 'PortfolioData_411136004.csv'

try:
    # 處理中文字元編碼問題
    try:
        df = pd.read_csv(filename, encoding='utf-8-sig')
    except:
        df = pd.read_csv(filename, encoding='big5')

    # 清除股價中的千分位逗號並轉換為數字格式
    df['收盤價(元)'] = df['收盤價(元)'].astype(str).str.replace(',', '').astype(float)

    # 將資料轉置 (Pivot) 成以日期為 Index，欄位為各股票代號的收盤價
    df_pivot = df.pivot(index='年月日', columns='代號', values='收盤價(元)')
    df_pivot.index = pd.to_datetime(df_pivot.index)
    df_pivot = df_pivot.sort_index()

    # 計算每日報酬率與年化數據 (假設一年 252 個交易日)
    returns = df_pivot.pct_change().dropna()
    mean_returns = returns.mean() * 252
    cov_matrix = returns.cov() * 252

except FileNotFoundError:
    print(f"錯誤：找不到檔案 '{filename}'，請確認是否已成功上傳至 Colab 根目錄。")
    raise

# --- 2. 定義最佳化函數 (尋找夏普值最大的切點投資組合) ---
rf = 0.01  # 假設無風險利率為 1%

def portfolio_performance(weights, mean_returns, cov_matrix):
    p_ret = np.sum(mean_returns * weights)
    p_std = np.sqrt(np.dot(weights.T, np.dot(cov_matrix, weights)))
    return p_std, p_ret

def neg_sharpe_ratio(weights, mean_returns, cov_matrix, rf):
    p_std, p_ret = portfolio_performance(weights, mean_returns, cov_matrix)
    # scipy.minimize 只能找最小值，所以將夏普值加上負號
    return -(p_ret - rf) / p_std

def get_tangency_portfolio(mean_ret, cov_mat):
    num_assets = len(mean_ret)
    args = (mean_ret, cov_mat, rf)
    constraints = ({'type': 'eq', 'fun': lambda x: np.sum(x) - 1})
    bounds = tuple((0.0, 1.0) for _ in range(num_assets))  # 限制權重在 0~1 之間 (不放空)

    result = sco.minimize(neg_sharpe_ratio, num_assets*[1./num_assets,], args=args,
                          method='SLSQP', bounds=bounds, constraints=constraints)
    return result

# --- 3. 執行最佳化運算 ---
# 任務一：3 檔資產 (2881, 2603, 2330)
assets_3 = [2881, 2603, 2330]
mean_ret_3 = mean_returns[assets_3]
cov_mat_3 = cov_matrix.loc[assets_3, assets_3]
opt_3 = get_tangency_portfolio(mean_ret_3, cov_mat_3)
std_3, ret_3 = portfolio_performance(opt_3.x, mean_ret_3, cov_mat_3)
sharpe_3 = (ret_3 - rf) / std_3

# 任務二：4 檔資產 (2881, 2603, 2330, 1702)
assets_4 = [2881, 2603, 2330, 1702]
mean_ret_4 = mean_returns[assets_4]
cov_mat_4 = cov_matrix.loc[assets_4, assets_4]
opt_4 = get_tangency_portfolio(mean_ret_4, cov_mat_4)
std_4, ret_4 = portfolio_performance(opt_4.x, mean_ret_4, cov_mat_4)
sharpe_4 = (ret_4 - rf) / std_4

# --- 4. 繪製圖表 ---
fig, axes = plt.subplots(1, 3, figsize=(18, 6))

# 子圖 1：績效長條圖比較
categories = ['Expected Return', 'Volatility (Risk)', 'Sharpe Ratio']
val_3 = [ret_3, std_3, sharpe_3]
val_4 = [ret_4, std_4, sharpe_4]
x = np.arange(len(categories))
width = 0.35

axes[0].bar(x - width/2, val_3, width, label='3 Assets (Markowitz)', color='darkblue')
axes[0].bar(x + width/2, val_4, width, label='4 Assets (+1702)', color='firebrick')
axes[0].set_ylabel('Value')
axes[0].set_title('Tangency Portfolio Performance Comparison', fontsize=14)
axes[0].set_xticks(x)
axes[0].set_xticklabels(categories)
axes[0].legend()

# 在長條圖上標示數值
for i, v in enumerate(val_3):
    axes[0].text(i - width/2, v + 0.05, f"{v:.2f}", ha='center', va='bottom', fontsize=11)
for i, v in enumerate(val_4):
    axes[0].text(i + width/2, v + 0.05, f"{v:.2f}", ha='center', va='bottom', fontsize=11)

# 設定 Y 軸上限避免文字被切掉
axes[0].set_ylim(0, max(max(val_3), max(val_4)) * 1.15)

# 子圖 2：三資產權重甜甜圈圖
colors_3 = ['#1f77b4', '#ff7f0e', '#2ca02c']
axes[1].pie(opt_3.x, labels=[str(a) for a in assets_3], autopct='%1.1f%%',
            startangle=90, colors=colors_3, wedgeprops=dict(width=0.4, edgecolor='w'))
axes[1].set_title('3 Assets Tangency Portfolio\nWeights', fontsize=14)

# 子圖 3：四資產權重甜甜圈圖
colors_4 = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728']
axes[2].pie(opt_4.x, labels=[str(a) for a in assets_4], autopct='%1.1f%%',
            startangle=90, colors=colors_4, wedgeprops=dict(width=0.4, edgecolor='w'))
axes[2].set_title('4 Assets Tangency Portfolio\nWeights', fontsize=14)

plt.tight_layout()
plt.show()
```

```
import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import scipy.optimize as sco

# 1. 自動下載 2024 年真實股價資料
tickers = ['2881.TW', '2603.TW', '2330.TW', '1702.TW', '2359.TW']
print("正在下載 2024 年股價資料...")
data = yf.download(tickers, start='2024-01-01', end='2024-12-31')['Close']

# 計算每日報酬與年化數據 (2024年)
returns = data.pct_change().dropna()
mean_returns = returns.mean() * 252
cov_matrix = returns.cov() * 252

def portfolio_volatility(weights, mean_returns, cov_matrix):
    return np.sqrt(np.dot(weights.T, np.dot(cov_matrix, weights)))

def get_hyperbola_frontier(mean_ret, cov_mat):
    num_assets = len(mean_ret)
    # 為了畫出誇張的延伸，把目標報酬率上下限拉得很開
    target_returns = np.linspace(mean_ret.min() - 0.5, mean_ret.max() + 0.5, 100)
    vols, rets = [], []
    for tr in target_returns:
        bounds = tuple((-1.0, 2.0) for _ in range(num_assets)) # 允許一定程度放空以畫出雙曲線
        constraints = ({'type': 'eq', 'fun': lambda x: np.sum(mean_ret * x) - tr},
                       {'type': 'eq', 'fun': lambda x: np.sum(x) - 1})
        result = sco.minimize(portfolio_volatility, num_assets*[1./num_assets,],
                              args=(mean_ret, cov_mat), method='SLSQP',
                              bounds=bounds, constraints=constraints)
        if result.success:
            vols.append(result.fun)
            rets.append(tr)
    min_vol_idx = np.argmin(vols)
    return vols, rets, vols[min_vol_idx], rets[min_vol_idx]

# --- 組合 A：您的防禦陣型 (加 1702) ---
assets_A = ['2881.TW', '2603.TW', '2330.TW', '1702.TW']
mean_ret_A = mean_returns[assets_A]
cov_mat_A = cov_matrix.loc[assets_A, assets_A]
vols_A, rets_A, mvp_v_A, mvp_r_A = get_hyperbola_frontier(mean_ret_A, cov_mat_A)

# --- 組合 B：組員的極端攻擊陣型 (加 2359 所羅門) ---
assets_B = ['2881.TW', '2603.TW', '2330.TW', '2359.TW']
mean_ret_B = mean_returns[assets_B]
cov_mat_B = cov_matrix.loc[assets_B, assets_B]
vols_B, rets_B, mvp_v_B, mvp_r_B = get_hyperbola_frontier(mean_ret_B, cov_mat_B)

# --- 繪圖比較 ---
plt.figure(figsize=(12, 8))

# 畫出您的防禦型雙曲線 (藍色)
plt.plot(vols_A, rets_A, color='blue', label='Defensive: 4 Assets (+1702 南僑)', linewidth=2.5)
plt.scatter(mvp_v_A, mvp_r_A, marker='*', color='blue', s=200, zorder=5)

# 畫出組員的攻擊型雙曲線 (紅色)
plt.plot(vols_B, rets_B, color='red', label='Aggressive: 4 Assets (+2359 所羅門)', linewidth=2.5)
plt.scatter(mvp_v_B, mvp_r_B, marker='*', color='red', s=200, zorder=5)

# 標示出所有單一股票的實際位置 (讓教授看清楚 2359 有多誇張)
for asset in tickers:
    vol = np.sqrt(cov_matrix.loc[asset, asset])
    ret = mean_returns[asset]
    plt.scatter(vol, ret, color='black', s=50, zorder=6)
    plt.annotate(asset.replace('.TW', ''), (vol, ret), xytext=(8, 5), textcoords='offset points', fontsize=12)

plt.title('2024 Efficient Frontier Clash: Defense (1702) vs Offense (2359)', fontsize=16)
plt.xlabel('Annualized Volatility (Risk, $\sigma$)', fontsize=14)
plt.ylabel('Annualized Returns (Expected Return, E(R))', fontsize=14)
plt.legend(fontsize=12, loc='upper left')
plt.grid(True, linestyle='--', alpha=0.7)

# 設定廣闊的座標軸範圍以容納右上角的怪物
plt.xlim(0, max(np.sqrt(np.diag(cov_matrix))) + 0.2)
plt.ylim(min(mean_returns) - 0.2, max(mean_returns) + 0.5)

plt.show()
```

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import scipy.optimize as sco

# 1. 讀取並準備資料
filename = 'PortfolioData_411136004.csv' # 請確認這個檔案在 Colab 左側目錄中

try:
    # 嘗試讀取 CSV 並處理可能的中文編碼問題
    try:
        df = pd.read_csv(filename, encoding='utf-8-sig')
    except:
        df = pd.read_csv(filename, encoding='big5')

    # 清除股價中的逗號並轉為數字
    df['收盤價(元)'] = df['收盤價(元)'].astype(str).str.replace(',', '').astype(float)

    # 轉置資料
    df_pivot = df.pivot(index='年月日', columns='代號', values='收盤價(元)')
    df_pivot.index = pd.to_datetime(df_pivot.index)
    df_pivot = df_pivot.sort_index()

    # 計算年化報酬率與共變異數矩陣
    returns = df_pivot.pct_change().dropna()
    mean_returns = returns.mean() * 252
    cov_matrix = returns.cov() * 252

except FileNotFoundError:
    print(f"錯誤：找不到檔案 '{filename}'，請確認是否已上傳至 Colab。")
    # 若無檔案，為確保能展示圖表格式，使用模擬數據
    print("將使用模擬數據繪製圖表...")
    days = 252
    mu = np.array([0.60, 2.50, 0.05, 0.10]) / days
    vols = np.array([0.25, 0.80, 0.20, 0.15]) / np.sqrt(days)
    corr = np.array([[1.0, 0.2, 0.4, 0.3], [0.2, 1.0, 0.1, 0.1], [0.4, 0.1, 1.0, 0.2], [0.3, 0.1, 0.2, 1.0]])
    cov = np.outer(vols, vols) * corr
    returns_sim = np.random.multivariate_normal(mu, cov, days)
    df_pivot = pd.DataFrame(returns_sim, columns=[2881, 2603, 2330, 1702])
    mean_returns = df_pivot.mean() * 252
    cov_matrix = df_pivot.cov() * 252

# 2. 定義計算前緣雙曲線的函數
def portfolio_volatility(weights, mean_returns, cov_matrix):
    return np.sqrt(np.dot(weights.T, np.dot(cov_matrix, weights)))

def get_hyperbola_frontier(mean_ret, cov_mat):
    num_assets = len(mean_ret)
    # 擴展範圍以畫出完整雙曲線
    target_returns = np.linspace(mean_ret.min() - 0.5, mean_ret.max() + 0.5, 100)
    vols = []
    rets = []
    for tr in target_returns:
        bounds = tuple((-2.0, 3.0) for _ in range(num_assets)) # 允許作空
        constraints = ({'type': 'eq', 'fun': lambda x: np.sum(mean_ret * x) - tr},
                       {'type': 'eq', 'fun': lambda x: np.sum(x) - 1})
        result = sco.minimize(portfolio_volatility, num_assets*[1./num_assets,],
                              args=(mean_ret, cov_mat), method='SLSQP',
                              bounds=bounds, constraints=constraints)
        if result.success:
            vols.append(result.fun)
            rets.append(tr)

    # 找出最小變異投資組合 (MVP)
    min_vol_idx = np.argmin(vols)
    return vols, rets, vols[min_vol_idx], rets[min_vol_idx]

# --- 任務 1：3 檔資產 (2881, 2603, 2330) ---
assets_3 = [2881, 2603, 2330]
mean_ret_3 = mean_returns[assets_3]
cov_mat_3 = cov_matrix.loc[assets_3, assets_3]
vols_3, rets_3, mvp_vol_3, mvp_ret_3 = get_hyperbola_frontier(mean_ret_3, cov_mat_3)

# --- 任務 2：4 檔資產 (2881, 2603, 2330, 1702) ---
assets_4 = [2881, 2603, 2330, 1702]
mean_ret_4 = mean_returns[assets_4]
cov_mat_4 = cov_matrix.loc[assets_4, assets_4]
vols_4, rets_4, mvp_vol_4, mvp_ret_4 = get_hyperbola_frontier(mean_ret_4, cov_mat_4)

# =======================================================
# 3. 繪製疊加雙曲線與 MVP 星號 (完美重現第三張圖格式)
# =======================================================
plt.figure(figsize=(10, 6))

# 畫出兩條雙曲線
plt.plot(vols_3, rets_3, color='darkblue', label='3 Assets Hyperbola (Markowitz)', linewidth=2)
plt.plot(vols_4, rets_4, color='firebrick', label='4 Assets Hyperbola (Black Zero-Beta CAPM)', linewidth=2)

# 標示最小變異投資組合 (MVP) 的星號
plt.scatter(mvp_vol_3, mvp_ret_3, marker='*', color='blue', s=150, label='MVP (3 Assets)', zorder=5)
plt.scatter(mvp_vol_4, mvp_ret_4, marker='*', color='red', s=150, label='MVP (4 Assets)', zorder=5)

# 設定標題與軸標籤 (使用 r'' 來避免 \s 的 SyntaxWarning)
plt.title('Efficient Frontier Hyperbolas Comparison', fontsize=14)
plt.xlabel(r'Annualized Volatility (Risk, $\sigma$)', fontsize=12)
plt.ylabel('Annualized Returns (Expected Return, E(R))', fontsize=12)

# 圖表美化
plt.legend(loc='upper right')
plt.grid(True, linestyle='--', alpha=0.7)

# 限制座標軸範圍，讓曲線看起來更像您截圖的比例
plt.xlim(0.05, max(max(vols_3), max(vols_4)) + 0.1)
# plt.ylim(-0.15, 0.45) # 您可以依據真實數據的高低點解開註解微調 y 軸

plt.show()
```
