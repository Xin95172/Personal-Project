```python
import pandas as pd
from google.colab import drive
import os

# 1. 掛載 Google Drive
drive.mount('/content/drive')

# 2. 設定要搜尋的檔案名稱
trading_file_name = "資料來源1_含file欄位_作業版.csv"  # 剛才產生的檔案
expiry_file_name = "資料來源3(作業).csv"              # 到期日原始檔

print("🔍 正在搜尋相關檔案...")

trading_path = None
expiry_path = None

# 自動搜尋檔案路徑
for root, dirs, files in os.walk('/content/drive/MyDrive/'):
    if trading_file_name in files:
        trading_path = os.path.join(root, trading_file_name)
    if expiry_file_name in files:
        expiry_path = os.path.join(root, expiry_file_name)
    # 如果全形括號的話也找一下
    elif "資料來源3（作業）.csv" in files:
        expiry_path = os.path.join(root, "資料來源3（作業）.csv")

# 3. 執行資料處理
if trading_path and expiry_path:
    print(f"✅ 找到交易檔：{trading_path}")
    print(f"✅ 找到到期日檔：{expiry_path}")

    # 讀取資料 (到期日檔無標題，需手動給欄位名)
    df_trading = pd.read_csv(trading_path)
    df_expiry = pd.read_csv(expiry_path, header=None)
    df_expiry.columns = ['最後結算日', '契約月份', '結算價']

    # A. 轉換日期格式
    df_trading['年月日_dt'] = pd.to_datetime(df_trading['年月日'])
    df_expiry['最後結算日_dt'] = pd.to_datetime(df_expiry['最後結算日'])

    # B. 篩選月選擇權 (排除名稱含 'W' 的週契約)
    df_monthly = df_expiry[~df_expiry['契約月份'].astype(str).str.contains('W')].copy()
    df_monthly = df_monthly.sort_values('最後結算日_dt')

    # C. 定義邏輯：尋找最近一個月合約且 Maturity >= 1
    def get_contract_info(trade_date):
        # 條件：最後結算日 - 交易日 >= 1
        valid = df_monthly[df_monthly['最後結算日_dt'] >= (trade_date + pd.Timedelta(days=1))]
        if not valid.empty:
            nearest = valid.iloc[0]
            contract = nearest['契約月份']
            expiry = nearest['最後結算日_dt']
            maturity = (expiry - trade_date).days
            return pd.Series([contract, expiry.strftime('%Y/%m/%d'), maturity])
        return pd.Series([None, None, None])

    # 執行比對
    print("⏳ 正在計算契約與到期天數...")
    df_trading[['Contract', 'ContractExpiryDate', 'Maturity']] = df_trading['年月日_dt'].apply(get_contract_info)

    # 4. 儲存結果
    output_path = os.path.join(os.path.dirname(trading_path), '資料來源1_含Contract資訊_作業版.csv')
    # 刪除暫存的日期欄位後存檔
    df_final = df_trading.drop(columns=['年月日_dt'])
    df_final.to_csv(output_path, index=False, encoding='utf-8-sig')

    print("-" * 30)
    print(f"🎉 處理成功！新檔案已儲存：\n{output_path}")
    display(df_final.head(10))
else:
    print("❌ 搜尋失敗，請確認這兩個檔案都在雲端硬碟中：")
    print(f"1. {trading_file_name}")
    print(f"2. {expiry_file_name}")
```
